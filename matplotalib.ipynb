```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x **2
y
```

```python
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
```

subplot

```python

plt.subplot(2, 2, 1)
plt.plot([1, 2, 3], [4, 5, 6])

plt.subplot(2, 2, 2)
plt.plot(x, y)

plt.subplot(2, 2, 3)
plt.plot([1, 2, 3], [2, 3, 4])

plt.subplot(2, 2, 4)
plt.plot(y, x)

plt.show()
```
